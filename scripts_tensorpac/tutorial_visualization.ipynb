# Visualization tutorial

The `functionalpac` package has several built-in visualization capabilities that may help intepreting the results. Being a package focused on the estimation of connectivity via PAC matrices 

## Types of visualization

We have two main types of visualization: the **matrix** visualization and the **head** visualization. 

- **Matrix visualization** are the traditional view of adjacency matrices, and the most straightforward. 

## Advanced head visualization

One can configure several aspects of the PAC connectivity using the BANDS directive.
