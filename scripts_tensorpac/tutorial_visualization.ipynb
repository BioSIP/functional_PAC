# Visualization tutorial

The `functionalpac` package has several built-in visualization capabilities that may help intepreting the results. Being a package focused on the estimation of connectivity via PAC matrices 

```python
from functionalpac.statistics import PermutationTesting, ClusterBasedPermutationTesting
```

## Types of visualization

We have two main types of visualization: the **matrix** visualization and the **head** visualization. 

- **Matrix visualization** are the traditional view of adjacency matrices, and the most straightforward. 

```python
import numpy as np 
from functionalpac.visualization import BandConnectivityPlotter
channels = ['Fp1', 'Fp2', 'F7', 'F3', 'Fz', 'F4', 'F8', 'FC5',
                             'FC1', 'FC2']

hipermatrix = np.random.randn(10, 10, 30, 40)
bands = {'Delta': (0.5, 4),
         'Theta': (4, 8),
                        'Alpha': (8, 12),
                        'Beta': (12, 25),
                        'Gamma': (25, 100),
                        'Epsilon': (100, 200)}
plotter = BandConnectivityPlotter(channels, bands)
```

```python
plotter.plot_connect_hipermatrix(hipermatrix)
```

```python
plotter.display_band_connectivity(hipermatrix)
```
